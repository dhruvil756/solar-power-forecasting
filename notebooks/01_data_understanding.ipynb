```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/Solar_Weather_Data_Processed.csv")
print(df.columns.tolist())
```

```python
print("Shape:", df.shape)
df.info()
df.head()
df.tail()
```

```python
time_col = [c for c in df.columns if "time" in c.lower() or "date" in c.lower()]
print("Possible time columns:", time_col)
df["timestamp"] = pd.to_datetime(df[time_col[0]])
df = df.sort_values("timestamp").reset_index(drop=True)
print("Start:", df["timestamp"].min())
print("End:  ", df["timestamp"].max())
print("Span (days):", (df["timestamp"].max() - df["timestamp"].min()).days)
print("Time step:", df["timestamp"].diff().value_counts().head())
```

```python
print(df.isna().sum())
print("Duplicate timestamps:", df["timestamp"].duplicated().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.describe().T
```

```python
target = "AC Output"
print("Rows with zero output:", (df[target] == 0).sum(), f"({(df[target] == 0).mean():.1%})")
print("Negative values:", (df[target] < 0).sum())
df[target].describe()
```

```python
df.columns = df.columns.str.strip()

weather_cols = ["GHI", "Temperature", "Relative Humidity", "Wind Speed", "Wind Direction"]
print("Weather columns:", weather_cols)
print("Missing per column:\n", df[weather_cols + ["AC Output"]].isna().sum())
print("Rows with both power and weather:", df[weather_cols + ["AC Output"]].notna().all(axis=1).sum())
```
