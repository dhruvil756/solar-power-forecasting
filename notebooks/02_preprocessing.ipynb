```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/Solar_Weather_Data_Processed.csv")
df.columns = df.columns.str.strip()
df["timestamp"] = pd.to_datetime(df["Timestamp"])
df = df.sort_values("timestamp").reset_index(drop=True)

# Inputs (X) and target (y)
feature_cols = [
    "GHI",
    "Temperature",
    "Relative Humidity",
    "Wind Speed",
    "Wind Direction",
    "Hour",
    "Month",
    "SZ Angle",
    "SA Angle",
]
target_col = "AC Output"

print("Number of features:", len(feature_cols))
print(df[feature_cols + [target_col]].describe().T[["mean", "min", "max"]])
```

```python
night = df[target_col] == 0
print("Zero-output rows:", night.sum(), f"({night.mean():.1%})")
print(df.loc[night, "Hour"].value_counts().sort_index())
```

```python
split_idx = int(len(df) * 0.8)
train = df.iloc[:split_idx]
test = df.iloc[split_idx:]

X_train, y_train = train[feature_cols], train[target_col]
X_test, y_test = test[feature_cols], test[target_col]

print("Train:", train["timestamp"].min(), "to", train["timestamp"].max(), f"({len(train)} rows)")
print("Test: ", test["timestamp"].min(), "to", test["timestamp"].max(), f"({len(test)} rows)")
```
